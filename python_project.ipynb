```python
class PVR:
    def bookticket(self, moviename, seats):
        print("PVR Cinemas")
        print(f"{seats} tickets confirmed for {moviename}")
        print("Payment and cancellation here")

    def cancelticket(self, ticket_id):
        print(f"Processing cancellation for ticket {ticket_id}")
        print("Amount will be refunded in 2 days")
```

```python
class Inox:
    def bookticket(self, moviename, seats):
        print("Inox Cinemas")
        print(f"{seats} tickets confirmed for {moviename}")
        print("Payment cancellation here")
        print("Food ordering available")

    def cancelticket(self, ticket_id):
        print("Cancellation not applicable")
```

```python
print("Welcome to BookMyShow")

mname = input("Movie name: ")
seats = int(input("How many tickets: "))

print("press 1 for PVR")
print("press 2 for Inox")

choice = int(input("choose: "))

theatre = None

if choice == 1:
    theatre = PVR()
elif choice == 2:
    theatre = Inox()
else:
    print("Invalid option")
    exit()

theatre.bookticket(mname, seats)
theatre.cancelticket(1234)
```
